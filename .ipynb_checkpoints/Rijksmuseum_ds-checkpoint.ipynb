```python
# ----------------------------------------------------------------------
# Process all CSV files
# ----------------------------------------------------------------------
input_dir = "processed_data/user_datasets_rijksmuseum"
output_dir = "processed_data/user_predicted_datasets_rijksmuseum"
os.makedirs(output_dir, exist_ok=True)


for file in glob.glob(os.path.join(input_dir, "*.csv")):
    df = pd.read_csv(file)
    
    # Use only the "with user info" model
    X, _ = prepare_data(df, config_with)
    X_filtered = {k: X[k] for k in input_names_with if k in X}
    pred = model_with.predict(X_filtered, verbose=0).flatten()
    
    # Add a single column, optionally replacing the original or keeping both
    df["predicted_engagement"] = pred   # new column
    
    # If you want to replace the original engagement_score (not recommended)
    # df["engagement_score"] = pred
    
    out_file = os.path.join(output_dir, os.path.basename(file))
    df.to_csv(out_file, index=False)

print("Finished.")
'''
for file in glob.glob(os.path.join(input_dir, "*.csv")):
    print(f"Processing {os.path.basename(file)}")
    
    df = pd.read_csv(file)
    
    # ---------- Scenario 1: LLM_enhanced ----------
    X_without, _ = prepare_data(df, config_without)
    # Filter to only the inputs the model expects (avoids extra keys)
    X_without_filtered = {k: X_without[k] for k in input_names_without if k in X_without}
    pred_without = model_without.predict(X_without_filtered, verbose=0).flatten()
    
    # ---------- Scenario 2: LLM_enhanced_and_user_info ----------
    X_with, _ = prepare_data(df, config_with)
    X_with_filtered = {k: X_with[k] for k in input_names_with if k in X_with}
    pred_with = model_with.predict(X_with_filtered, verbose=0).flatten()
    
    # ---------- Store predictions and ranks ----------
    df["pred_engagement_without_user"] = pred_without
    df["pred_engagement_with_user"] = pred_with
    
    df["rank_without_user"] = df["pred_engagement_without_user"].rank(
        ascending=False, method="dense"
    )
    df["rank_with_user"] = df["pred_engagement_with_user"].rank(
        ascending=False, method="dense"
    )
    
    # Save the updated DataFrame (optional)
    out_file = os.path.join(output_dir, os.path.basename(file))
    df.to_csv(out_file, index=False)
    print(f"Saved predictions to {out_file}")

print("Finished.")
'''
```
