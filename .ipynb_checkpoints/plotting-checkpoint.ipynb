```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error, ndcg_score, dcg_score
```

```python
def compute_ndcg_at_k(y_true, y_score, k):
    # DCG@k using sklearn
    dcg = dcg_score([y_true], [y_score], k=k)
    # Ideal DCG
    ideal = np.sort(y_true)[::-1][:k]
    idcg = dcg_score([ideal], [ideal], k=k)
    return dcg / idcg if idcg > 0 else 0.0


#Compare regression and ranking metrics across different scenarios.
def compare_metrics_and_plotting(df_dict, threshold=0.0, k_values=[5, 10, 20], show_extreme_users=True, n_extreme=5):
    
    results = {}
    per_user_metrics = {}

    for name, df in df_dict.items():
        y_true = df["true_engagement"].values
        y_pred = df["predicted_engagement"].values

        # Regression metrics
        mae = mean_absolute_error(y_true, y_pred)
        rmse = np.sqrt(mean_squared_error(y_true, y_pred))

        # --- Global NDCG (entire dataset as one query) ---
        y_true_bin = (y_true > threshold).astype(int)
        global_ndcg = ndcg_score([y_true_bin], [y_pred])   # uses all items

        # Per‑user ranking metrics
        user_rows = []
        if "user_id_encoded" in df.columns:
            for user_id, group in df.groupby("user_id_encoded"):
                if len(group) < 2:
                    continue
                y_true_bin_user = (group["true_engagement"].values > threshold).astype(int)
                if y_true_bin_user.sum() == 0:    # skip users with no positive items
                    continue
                y_pred_user = group["predicted_engagement"].values

                row = {'user_id': user_id}
                for k in k_values:
                    # NDCG@k using sklearn-based helper
                    ndcg = compute_ndcg_at_k(y_true_bin_user, y_pred_user, k)
                    row[f'NDCG@{k}'] = ndcg

                    # Precision, Recall, MAP (manual, but robust)
                    order = np.argsort(-y_pred_user)
                    y_true_sorted = y_true_bin_user[order]
                    total_pos = y_true_bin_user.sum()
                    effective_k = min(k, len(y_true_sorted))
                    relevant_in_topk = np.sum(y_true_sorted[:k])

                    # Precision@k
                    row[f'P@{k}'] = relevant_in_topk / effective_k if effective_k > 0 else 0.0

                    # Recall@k
                    row[f'R@{k}'] = relevant_in_topk / total_pos if total_pos > 0 else 0.0

                    # MAP@k
                    ap = 0.0
                    hits = 0
                    for i in range(effective_k):
                        if y_true_sorted[i] == 1:
                            hits += 1
                            ap += hits / (i + 1)
                    denom = min(total_pos, effective_k)
                    row[f'MAP@{k}'] = ap / denom if denom > 0 else 0.0

                user_rows.append(row)

            df_users = pd.DataFrame(user_rows)
            per_user_metrics[name] = df_users

            # Aggregate over users (mean)
            agg = {"MAE": mae, "RMSE": rmse, "Global_NDCG": global_ndcg}
            for k in k_values:
                agg[f'NDCG@{k}'] = df_users[f'NDCG@{k}'].mean()
                agg[f'P@{k}'] = df_users[f'P@{k}'].mean()
                agg[f'R@{k}'] = df_users[f'R@{k}'].mean()
                agg[f'MAP@{k}'] = df_users[f'MAP@{k}'].mean()
            results[name] = agg
        else:
            print(f"Warning: 'user_id_encoded' missing in {name}. Skipping per‑user ranking metrics.")
            results[name] = {"MAE": mae, "RMSE": rmse, "Global_NDCG": global_ndcg}
            per_user_metrics[name] = pd.DataFrame()

    #############################################################
    # Boxplots: NDCG, MAP, Precision, Recall (per user, across k)
    if any(not df.empty for df in per_user_metrics.values()):
        n_scenarios = len(df_dict)
        n_k = len(k_values)
        metrics = ['NDCG', 'MAP', 'Precision', 'Recall']   # displayed as NDCG@k, MAP@k, P@k, R@k

        fig, axes = plt.subplots(2, 2, figsize=(14, 10))
        axes = axes.flatten()

        for ax_idx, metric in enumerate(metrics):
            if metric == 'MAP':
                col_template = 'MAP@{k}'
            elif metric == 'NDCG':
                col_template = 'NDCG@{k}'
            elif metric == 'Precision':
                col_template = 'P@{k}'
            else:   # Recall
                col_template = 'R@{k}'

            all_scores = []
            all_labels = []
            positions = []
            for i, k in enumerate(k_values):
                for j, name in enumerate(df_dict.keys()):
                    df_u = per_user_metrics.get(name)
                    col_name = col_template.format(k=k)
                    if df_u is not None and not df_u.empty and col_name in df_u.columns:
                        scores = df_u[col_name].dropna().values
                        if len(scores) > 0:
                            all_scores.append(scores)
                            all_labels.append(f"{name}\n{col_name}")
                            positions.append(i * (n_scenarios + 0.5) + j)

            if all_scores:
                bp = axes[ax_idx].boxplot(all_scores, positions=positions, widths=0.35, patch_artist=True)
                # Style medians black
                for median_line in bp['medians']:
                    median_line.set_color('black')
                # Color boxes
                colors = ['lightblue', 'lightgreen', 'lightcoral', 'lightyellow', 'lightpink']
                for idx, patch in enumerate(bp['boxes']):
                    patch.set_facecolor(colors[idx % len(colors)])

                axes[ax_idx].set_xticks(positions)
                axes[ax_idx].set_xticklabels(all_labels, rotation=45, ha='right')
                axes[ax_idx].set_ylabel(f'{metric} score')
                axes[ax_idx].set_title(f'{metric}@k per User')
                axes[ax_idx].grid(axis='y', alpha=0.3)
            else:
                axes[ax_idx].text(0.5, 0.5, f'No data for {metric}', ha='center', va='center')
                axes[ax_idx].set_title(f'{metric}@k')

        plt.suptitle('User‑level Ranking Metrics Comparison', fontsize=16)
        plt.tight_layout(rect=[0, 0, 1, 0.96])
        plt.savefig('models/metrics_boxplots.png', dpi=400, bbox_inches='tight')
        plt.show()
    else:
        print("No per‑user data available – skipping boxplots.")

    # Print main comparison table
    if results:
        print("\n" + "=" * 120)
        print("Performance Comparison".center(120))
        print("=" * 120)
        df_table = pd.DataFrame(results).T.round(4)
        # Order columns: MAE, RMSE, Global_NDCG, then per‑user metrics per k
        ordered_cols = ['MAE', 'RMSE', 'Global_NDCG']
        for k in k_values:
            ordered_cols += [f'NDCG@{k}', f'P@{k}', f'R@{k}', f'MAP@{k}']
        # Keep only existing columns (safety)
        ordered_cols = [col for col in ordered_cols if col in df_table.columns]
        df_table = df_table[ordered_cols]
        print(df_table.to_string())
        print("=" * 120)

    #############################################################
    # Extreme‑user removal analysis (based on average NDCG over k_values)
    for name, df_u in per_user_metrics.items():
        if df_u.empty:
            continue

        # Average NDCG across all k for each user
        ndcg_cols = [f'NDCG@{k}' for k in k_values]
        df_u['avg_ndcg'] = df_u[ndcg_cols].mean(axis=1)

        n_users = len(df_u)
        n_remove = min(n_extreme, n_users - 1)
        if n_remove == 0:
            print(f"\nSkipping extreme‑user analysis for '{name}' – too few users ({n_users})")
            continue

        best_ids = df_u.nlargest(n_remove, 'avg_ndcg')['user_id'].tolist()
        worst_ids = df_u.nsmallest(n_remove, 'avg_ndcg')['user_id'].tolist()

        subsets = {
            "Original": None,
            f"Remove {n_remove} best": best_ids,
            f"Remove {n_remove} worst": worst_ids,
            f"Remove {n_remove} best & worst": best_ids + worst_ids
        }

        original_df = df_dict[name]
        all_user_ids = set(df_u['user_id'])

        subset_results = {}
        for subset_name, remove_ids in subsets.items():
            if remove_ids is None:
                user_subset = all_user_ids
            else:
                user_subset = all_user_ids - set(remove_ids)

            if len(user_subset) == 0:
                print(f"  {subset_name}: no users left – skipping")
                continue

            mask = original_df['user_id_encoded'].isin(user_subset)
            df_sub = original_df[mask]
            y_true_sub = df_sub['true_engagement'].values
            y_pred_sub = df_sub['predicted_engagement'].values

            # Regression on subset
            mae_sub = mean_absolute_error(y_true_sub, y_pred_sub)
            rmse_sub = np.sqrt(mean_squared_error(y_true_sub, y_pred_sub))

            # Global NDCG on subset (full list)
            y_true_bin_sub = (y_true_sub > threshold).astype(int)
            global_ndcg_sub = ndcg_score([y_true_bin_sub], [y_pred_sub])

            # Per‑user averages on the subset
            df_u_sub = df_u[df_u['user_id'].isin(user_subset)]
            metrics_row = {"MAE": mae_sub, "RMSE": rmse_sub, "Global_NDCG": global_ndcg_sub}
            for k in k_values:
                metrics_row[f'NDCG@{k}'] = df_u_sub[f'NDCG@{k}'].mean()
                metrics_row[f'P@{k}'] = df_u_sub[f'P@{k}'].mean()
                metrics_row[f'R@{k}'] = df_u_sub[f'R@{k}'].mean()
                metrics_row[f'MAP@{k}'] = df_u_sub[f'MAP@{k}'].mean()
            subset_results[subset_name] = metrics_row

        if subset_results:
            print(f"\n--- {name}: Effect of removing extreme users (by avg NDCG) ---")
            df_effect = pd.DataFrame(subset_results).T.round(4)
            ordered_cols = ['MAE', 'RMSE', 'Global_NDCG']
            for k in k_values:
                ordered_cols += [f'NDCG@{k}', f'P@{k}', f'R@{k}', f'MAP@{k}']
            ordered_cols = [col for col in ordered_cols if col in df_effect.columns]
            df_effect = df_effect[ordered_cols]
            print(df_effect.to_string())
            print("-" * 100)

    # Show extreme users 
    if show_extreme_users and per_user_metrics:
        for name, df_u in per_user_metrics.items():
            if df_u.empty:
                continue
            ndcg_cols = [f'NDCG@{k}' for k in k_values]
            df_u['avg_ndcg'] = df_u[ndcg_cols].mean(axis=1)
            df_sorted = df_u.sort_values('avg_ndcg', ascending=False)
            best = df_sorted.head(n_extreme)
            worst = df_sorted.tail(n_extreme).iloc[::-1]
            print(f"\n--- {name} ---")
            print(f"Top {n_extreme} users (highest avg NDCG):")
            print(best[['user_id', 'avg_ndcg'] + ndcg_cols].to_string(index=False))
            print(f"\nBottom {n_extreme} users (lowest avg NDCG):")
            print(worst[['user_id', 'avg_ndcg'] + ndcg_cols].to_string(index=False))

    return results, per_user_metrics


if __name__ == "__main__":
    # Load your prediction CSVs
    df1 = pd.read_csv("models/test_predictions_LLM_enhanced.csv")
    df2 = pd.read_csv("models/test_predictions_LLM_enhanced_and_user_info.csv")

    results, per_user = compare_metrics_and_plotting(
        df_dict={"LLM features": df1, "LLM features + user info": df2},
        threshold=0.0,
        k_values=[5, 10, 20],
        show_extreme_users=True,
        n_extreme=5
    )
```
